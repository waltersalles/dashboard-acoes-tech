```python
!pip install dash

import yfinance as yf
import pandas as pd
import plotly.graph_objects as go
from dash import Dash, dcc, html, Input, Output

# Dicionário das ações
acoes = {
    'GOOGL': 'Google',
    'AAPL': 'Apple',
    'MSFT': 'Microsoft',
    'NVDA': 'Nvidia'
}

# Inicializa o app
app = Dash(__name__)
app.title = "Dashboard Ações Tech"

app.layout = html.Div([
    html.H1("Dashboard - Ações Tech", style={'textAlign': 'center'}),

    html.Div([
        html.Label("Selecione as ações:"),
        dcc.Checklist(
            id='seletor-acoes',
            options=[{'label': nome, 'value': ticker} for ticker, nome in acoes.items()],
            value=list(acoes.keys()),
            inline=True
        )
    ], style={'margin': '20px'}),

    html.Div([
        html.Label("Selecione o intervalo de datas:"),
        dcc.DatePickerRange(
            id='intervalo-datas',
            start_date=pd.to_datetime("2025-01-01"),
            end_date=pd.to_datetime("2025-03-31"),
            display_format='DD/MM/YYYY'
        ),
    ], style={'textAlign': 'center', 'marginBottom': 30}),

    html.Div(id='graficos-acoes')
])

# Callback para atualizar os gráficos
@app.callback(
    Output('graficos-acoes', 'children'),
    Input('seletor-acoes', 'value'),
    Input('intervalo-datas', 'start_date'),
    Input('intervalo-datas', 'end_date')
)
def atualizar_graficos(selecionadas, data_inicio, data_fim):
    graficos = []

    if not selecionadas:
        return html.P("Selecione ao menos uma ação para visualizar os dados.", style={'color': 'red', 'textAlign': 'center'})

    for ticker in selecionadas:
        nome = acoes[ticker]
        df = yf.download(ticker, start=data_inicio, end=data_fim)
        if df.empty:
            continue
        df.reset_index(inplace=True)

        fig = go.Figure()
        fig.add_trace(go.Scatter(
            x=df['Date'],
            y=df['Close'],
            mode='lines+markers',
            name=nome
        ))
        fig.update_layout(
            title=f'{nome} ({ticker})',
            xaxis_title='Data',
            yaxis_title='Preço de Fechamento (USD)',
            template='plotly_dark'
        )
        graficos.append(dcc.Graph(figure=fig))

    return graficos

# Executa o servidor
if __name__ == '__main__':
    app.run(debug=True)
```
